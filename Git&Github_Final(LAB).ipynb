```python
from Bio.Seq import Seq

dna = Seq("ATGGCGCGCTTA")
gc_content = (dna.count("G") + dna.count("C")) / len(dna) * 100
print("DNA Sequence:", dna)
print(f"GC Content: {round(gc_content, 2)}%")
```

```python
from Bio.Seq import Seq

dna = Seq("ATGGCGCGCTTA")
at_content = (dna.count("A") + dna.count("T")) / len(dna) * 100
print("DNA Sequence:", dna)
print(f"AT Content: {round(at_content, 2)}%")
```

```python
from Bio.Seq import Seq

dna = Seq("ATGGCGCGCTTA")

# Length
seq_length = len(dna)

# Reverse Complement
rev_comp = dna.reverse_complement()

# Translation
protein_seq = dna.translate()

print(f"Sequence Length: {seq_length} bp")
print(f"Reverse Complement: {rev_comp}")
print(f"Translated Protein: {protein_seq}")
```
